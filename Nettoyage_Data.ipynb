```python
for col in ['Age' ,"Salaire_MAD", "Note_Performance", "Satisfaction_Score"]:
    df[col] = df[col].fillna(df[col].median())
for col in ['Email']:
    df[col] = df[col].fillna('inconnu')
for col in ['Departement', 'Ville' ,'Type_Contrat']:
    df[col] = df[col].fillna(statistics.mode(df[col])
                             )
```

```python
def outliers(serie):
    q1 , q3 = np.quantile(serie , 0.25) , np.quantile(serie , 0.75)
    iqr = q3 - q1
    borne_inf , borne_sup = q1 - 1.5 * iqr , q3 + 1.5 * iqr
    outlier = (serie > borne_sup) | (serie < borne_inf)
    return outlier 
```

```python
outlier1 = outliers(df['Age'])
df.loc[outlier1, 'Age'] = np.nan
median_age = round(df['Age'].median())
df['Age'] = df['Age'].fillna(median_age)
```

```python
outlier2 = outliers(df['Salaire_MAD'])
df.loc[outlier2 , 'Salaire_MAD'] = np.nan
median = round(df['Salaire_MAD'].median())
df['Salaire_MAD'] = df['Salaire_MAD'].fillna(median)
```

```python
outlier3 = outliers(df['Note_Performance'])
df.loc[outlier3 , 'Note_Performance'] = np.nan
median = round(df['Note_Performance'].median())
df['Note_Performance'] = df['Note_Performance'].fillna(median)
```

```python
outlier4 = outliers(df['Heures_Sup_Mensuelles'])
df.loc[outlier4 , 'Heures_Sup_Mensuelles'] = np.nan
median = round(df['Heures_Sup_Mensuelles'].median())
df['Heures_Sup_Mensuelles'] = df['Heures_Sup_Mensuelles'].fillna(median)
```
