```python
valeurs_num = df.select_dtypes(include= "number")
valeurs_cat = df.select_dtypes(include= "string")
for i in valeurs_num:
    manquante_num = pd.isnull(valeurs_num[i]).sum()
    manquante_num/len(valeurs_num[i])
    print(i ,":" ,manquante_num, "%")
for j in valeurs_cat:
    manquante_num = pd.isnull(valeurs_cat[j]).sum()
    manquante_num/len(valeurs_cat[j])
    print(j ,":" ,manquante_num, "%")
```

```python
for i in valeurs_num:
    min = valeurs_num[i].min()
    max = valeurs_num[i].max()
    etendu = max - min
    print("min de " ,i ," est :", min ,"le max de", i ,"est :", max ,'etendue est ', etendu)
    print()
```

```python
doublans_num = valeurs_num.duplicated().sum()
doublans_cat = valeurs_cat.duplicated().sum()
print('les valeurs doublans numerique :  ' ,doublans_num)
print('les valeurs doubalans categorielles :  ' ,doublans_cat)
```
